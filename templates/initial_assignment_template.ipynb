## Task 1 — Clean and split the text

Write a function `tokenize(text)` that:

- converts the text to lowercase;
- removes `.`, `,`, `!`, `?`, `;`, and `:`;
- splits the text into words; and
- returns a list containing the words.

Test the function with `sample_text` and print the resulting list. Do not use external libraries.

```python
# Write your Task 1 solution here.
import string
def tokenize(text): 
    new_text = text.lower()
    new_text = ''.join([char for char in new_text if char not in string.punctuation])
    new_text = new_text.split()
    return new_text
result_tokenized = tokenize(sample_text)
print(result_tokenized)
```

**Task 1 explanation:** My function starts with lowering everything into a lower character so there will be no capital letter at all. Afterwards with the string function it would remove any possible punctuations. Eventually by using the .split() it would split everything into words

## Task 2 — Count word frequencies

Write a function `word_frequencies(tokens)` that receives the list from Task 1 and returns a dictionary. Each word should be a key and its number of occurrences should be the value.

Use a loop and a dictionary; do not use `collections.Counter`. Print the returned dictionary.

```python
# Write your Task 2 solution here.
def word_frequencies(tokens):
    dct = {}
    for word in tokens:
        if word not in dct:
            dct[word] = 1
        else:
            dct[word] += 1
    return dct
frequencies = word_frequencies(result_tokenized)
print(frequencies)
```

**Task 2 explanation:** This function first of all initializes an empty dictionary. Afterwards, iterates over the cleaned list of items to check whether they are already exist in the dictionary or not. In case they are already in the dictionary, it assignes the value of +1 otherwise sets the value of it to 1

## Task 3 — Produce a summary

Write a function `text_summary(tokens, frequencies)` that returns a dictionary with these four fields:

- `total_words`;
- `unique_words`;
- `average_word_length`, rounded to two decimal places; and
- `most_common_word`.

If two or more words have the highest frequency, choose the alphabetically first one. Print your summary dictionary.

```python
# Write your Task 3 solution here.

def text_summary(tokens, frequencies): 
    dct = dict()
    length_words = 0
    dct["total_words"] = len(tokens)
    dct["unique_words"] = list(frequencies.keys())
    for word in dct["unique_words"]:
        length_words += len(word)
    dct["average_word_length"] = round(length_words / len(dct["unique_words"]),2)
    dct["most_common_word"] = max(frequencies, key=frequencies.get)
    return dct
summarized = text_summary(result_tokenized, frequencies)
print(summarized)
```

**Task 3 explanation:** Initialized the empty dictionary, calculated the length of the tokens, calculated the total length and divided by the number of words to see the average length of the words. Then found the most_common_word by using the built in max function. Returned to the dictionary

## Task 4 — Find a word

Write a function `find_word(tokens, query)` that returns a dictionary with:

- `count`: the number of times `query` appears; and
- `positions`: a list of the zero-based positions where it appears.

Convert `query` to lowercase before searching. Test the function with at least `"data"` and one word that does not occur in the text.

```python
# Write your Task 4 solution here.
def find_word(tokens, query):
    query = query.lower()
    count = 0
    positions = []
    for i, word in enumerate(tokens):
        if word == query:
            positions.append(i)

    dct = {'count': len(positions),
          'positions': positions}
    return dct
    
print(find_word(result_tokenized, "data"))
print(find_word(result_tokenized, "nonsense"))
    
```
